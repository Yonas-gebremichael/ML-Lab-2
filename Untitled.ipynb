```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
n = 100

study_hours = np.random.normal(5, 2, n)
attendance = np.random.normal(85, 10, n)
previous_mark = np.random.normal(70, 15, n)

final_score = (
    0.5 * study_hours
    + 0.3 * attendance
    + 0.2 * previous_mark
    + np.random.normal(0, 5, n)
)

df = pd.DataFrame({
    "study_hours": study_hours,
    "attendance": attendance,
    "previous_mark": previous_mark,
    "final_score": final_score
})

df.loc[0:5, "study_hours"] = np.nan
df.loc[99, "study_hours"] = 50
df.loc[98, "attendance"] = -10

display(df)
```

```python
# Part A: Inspect and Clean
# 1. Remove row 99 (50 hours outlier) and row 98 (-10 attendance error)
df = df.drop([98, 99]) 

# Fill the missing (NaN) study hours with the median value
df['study_hours'] = df['study_hours'].fillna(df['study_hours'].median())

# 2. Print matrix details
print(f"Shape of feature matrix: {df.shape}")
print("Numerical columns: study_hours, attendance, previous_mark, final_score")
print("Target variable: final_score")
```

```python
# Part B: Summarize
# 1. Calculate mean, median, standard deviation
summary = df.agg(['mean', 'median', 'std'])
print(summary)

# 2. Three Observations for your report:
print("\nObservations:")
print("- The mean and median for 'attendance' are very close, suggesting a symmetrical distribution.")
print("- The standard deviation of 'previous_mark' is the highest, indicating wide variance in past performance.")
print("- 'study_hours' has a low standard deviation, showing most students study roughly the same amount.")
```

```python
import matplotlib.pyplot as plt

# Part C: Visualize
plt.figure(figsize=(10, 4))

# 1. Histogram for final_score
plt.subplot(1, 2, 1)
plt.hist(df['final_score'], bins=15, color='blue', edgecolor='black')
plt.title('Distribution of Final Scores')
plt.xlabel('Final Score')
plt.ylabel('Frequency')

# 2. Scatter plot: study_hours vs final_score
plt.subplot(1, 2, 2)
plt.scatter(df['study_hours'], df['final_score'], alpha=0.7, color='green')
plt.title('Study Hours vs Final Score')
plt.xlabel('Study Hours')
plt.ylabel('Final Score')

plt.tight_layout()
plt.show()

# Describe the trend for your report
print("Trend Description: There is a positive linear trend; as study hours increase, the final score generally tends to increase.")
```

```python
# Part D: Correlation and Causality
# 1. Pearson correlation
correlation = df['study_hours'].corr(df['final_score'])
print(f"Pearson correlation (study_hours vs final_score): {correlation:.2f}")

# 2. Critical Thinking (For your report):
print("\nCritical Thinking Answers:")
print("- Does 10 hours of study guarantee a score? No. Correlation only shows a general trend, not a strict mathematical guarantee.")
print("- Potential 'lurking' variable: Baseline knowledge (like 'previous_mark') or natural aptitude. A student who already knows the material might study less but score higher.")
```

```python
import numpy as np

# Part E: Vectors and Dot Product
# 1. Select one student row (index 0) excluding the target final_score
student_vector = df.loc[0, ['study_hours', 'attendance', 'previous_mark']].values

# 2. Define weights and bias
w = np.array([0.5, 0.2, 0.4])
b = 5

# 3. Calculate y = w dot x + b
y = np.dot(w, student_vector) + b

print(f"Student features (x): {student_vector}")
print(f"Weights (w): {w}, Bias (b): {b}")
print(f"Predicted score (y): {y:.2f}")

# 4. Conceptual Answers (For your report):
print("\nConceptual Answers:")
print("- 'w' represents the weight or importance of each feature. It tells the model how much impact study hours vs attendance has on the score.")
print("- 'b' represents the baseline score a student would get if all their features (study hours, attendance, etc.) were exactly zero.")
```

```python
# Part F: Scaling
# 1. Standardize study_hours and previous_mark
df['study_hours_scaled'] = (df['study_hours'] - df['study_hours'].mean()) / df['study_hours'].std()
df['previous_mark_scaled'] = (df['previous_mark'] - df['previous_mark'].mean()) / df['previous_mark'].std()

print("First 3 rows of scaled features:")
print(df[['study_hours_scaled', 'previous_mark_scaled']].head(3))

# 2. Conceptual Answer (For your report):
print("\nConceptual Answer:")
print("- Distance-based algorithms (like KNN) calculate physical distance between numbers. Because 'previous_mark' has a much larger range (e.g., 50-100) than 'study_hours' (e.g., 0-10), an unscaled model would treat 'previous_mark' as massively more important just because the raw numbers are bigger. Scaling puts both features on a level playing field.")
```
