```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

df = pd.read_csv("lab_ticket_resolution.csv")
print(df.shape)
print(df.dtypes)
print(df.describe(include="all"))
print(df.isna().sum())
```

```python
# 1. Finish Task 1: Show the first 5 rows
print("--- First 5 Rows ---")
print(df.head(5))

# 2. Task 2: Define y (Label) and X (Features)
y = df["resolution_hours"]

# Drop ticket_id (just an ID), resolution_hours (the target), and lines_changed (Data Leakage!)
X = df.drop(columns=["ticket_id", "resolution_hours", "lines_changed"])

print("\n--- Features we are actually using (X) ---")
print(X.columns.tolist())

# 3. Task 3: Split into 80% Training and 20% Testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print(f"\nTraining rows: {X_train.shape[0]}, Testing rows: {X_test.shape[0]}")
```

```python
# Helper function to calculate and print our 3 metrics cleanly
def print_metrics(model_name, y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    print(f"--- {model_name} ---")
    print(f"MAE:  {mae:.3f} hours")
    print(f"RMSE: {rmse:.3f} hours")
    print(f"R²:   {r2:.3f}\n")

# Task 4: Fit DummyRegressor baseline (predicts the simple average every time)
dummy = DummyRegressor(strategy="mean")
dummy.fit(X_train, y_train)
y_pred_dummy = dummy.predict(X_test)
print_metrics("Task 4: Dummy Baseline (Mean)", y_test, y_pred_dummy)

# Task 5: Fit Linear Regression on selected features
lin_reg = LinearRegression()
lin_reg.fit(X_train, y_train)
y_pred_lin = lin_reg.predict(X_test)
print_metrics("Task 5: Linear Regression", y_test, y_pred_lin)
```

```python
# List to store the error comparison of the 4 models
results = []

def test_model(name, model):
    # 1. Model Train (Fit)
    model.fit(X_train, y_train)
    
    # 2. Predict on test data
    preds = model.predict(X_test)
    
    # 3. Calculate the 3 Errors / Metrics
    mae = mean_absolute_error(y_test, preds)             # Mean Absolute Error
    rmse = np.sqrt(mean_squared_error(y_test, preds))    # Root Mean Square Error
    r2 = r2_score(y_test, preds)                         # R² (Residual Coefficient)
    
    results.append({
        "Model": name,
        "MAE": round(mae, 3),
        "RMSE": round(rmse, 3),
        "R2_Score": round(r2, 3)
    })

# 1. Dummy Regressor (Baseline)
dummy = DummyRegressor(strategy="mean")
test_model("1. Dummy (Mean)", dummy)

# 2. Linear Regression (Degree 1)
lin_deg1 = LinearRegression()
test_model("2. Linear (Degree 1)", lin_deg1)

# 3. Polynomial Regression (Degree 2)
poly_deg2 = make_pipeline(PolynomialFeatures(degree=2), StandardScaler(), LinearRegression())
test_model("3. Polynomial (Degree 2)", poly_deg2)

# 4. Polynomial Regression (Degree 3)
poly_deg3 = make_pipeline(PolynomialFeatures(degree=3), StandardScaler(), LinearRegression())
test_model("4. Polynomial (Degree 3)", poly_deg3)

# Print the final comparison table of the 4 models
comparison_df = pd.DataFrame(results)
print("=== COMPARISON OF THE 4 MODELS ===")
print(comparison_df.to_string(index=False))
```

```python
import matplotlib.pyplot as plt

# Get predictions from the models we already fitted
preds_deg1 = lin_deg1.predict(X_test)
preds_deg2 = poly_deg2.predict(X_test)

# Calculate residuals (Actual Hours - Predicted Hours)
residuals_deg1 = y_test - preds_deg1
residuals_deg2 = y_test - preds_deg2

# Create a 2x2 grid of plots
fig, axes = plt.subplots(2, 2, figsize=(12, 9))

# 1. Linear (Degree 1): Observed vs Predicted
axes[0, 0].scatter(y_test, preds_deg1, alpha=0.7, color='blue')
axes[0, 0].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
axes[0, 0].set_title("1. Linear (Deg 1): Observed vs Predicted")
axes[0, 0].set_xlabel("Actual Resolution Hours (y_test)")
axes[0, 0].set_ylabel("Predicted Hours")

# 2. Linear (Degree 1): Residual Plot
axes[0, 1].scatter(preds_deg1, residuals_deg1, alpha=0.7, color='blue')
axes[0, 1].axhline(0, color='r', linestyle='--', lw=2)
axes[0, 1].set_title("2. Linear (Deg 1): Residual Plot")
axes[0, 1].set_xlabel("Predicted Hours")
axes[0, 1].set_ylabel("Residual Error (Actual - Predicted)")

# 3. Polynomial (Degree 2): Observed vs Predicted
axes[1, 0].scatter(y_test, preds_deg2, alpha=0.7, color='green')
axes[1, 0].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
axes[1, 0].set_title("3. Polynomial (Deg 2): Observed vs Predicted")
axes[1, 0].set_xlabel("Actual Resolution Hours (y_test)")
axes[1, 0].set_ylabel("Predicted Hours")

# 4. Polynomial (Degree 2): Residual Plot
axes[1, 1].scatter(preds_deg2, residuals_deg2, alpha=0.7, color='green')
axes[1, 1].axhline(0, color='r', linestyle='--', lw=2)
axes[1, 1].set_title("4. Polynomial (Deg 2): Residual Plot")
axes[1, 1].set_xlabel("Predicted Hours")
axes[1, 1].set_ylabel("Residual Error (Actual - Predicted)")

plt.tight_layout()
plt.show()
```

```python
from sklearn.linear_model import RidgeCV, LassoCV

# --- TASK 6: Cross-Validation on Training Set (Degree 2 vs Degree 3) ---
kf = KFold(n_splits=5, shuffle=True, random_state=42)

cv_deg2 = -cross_val_score(poly_deg2, X_train, y_train, scoring="neg_root_mean_squared_error", cv=kf).mean()
cv_deg3 = -cross_val_score(poly_deg3, X_train, y_train, scoring="neg_root_mean_squared_error", cv=kf).mean()

print("=== TASK 6: 5-Fold Cross-Validation on Training Data ===")
print(f"Polynomial Degree 2 CV RMSE: {cv_deg2:.3f} hours")
print(f"Polynomial Degree 3 CV RMSE: {cv_deg3:.3f} hours\n")

# --- TASK 7: Ridge and Lasso with Scaling & Alpha Selection ---
alphas_to_test = [0.01, 0.1, 1.0, 10.0, 100.0]

# Fit Ridge
ridge_pipe = make_pipeline(StandardScaler(), RidgeCV(alphas=alphas_to_test, cv=kf))
ridge_pipe.fit(X_train, y_train)
ridge_preds = ridge_pipe.predict(X_test)
best_ridge_alpha = ridge_pipe.named_steps['ridgecv'].alpha_

# Fit Lasso
lasso_pipe = make_pipeline(StandardScaler(), LassoCV(alphas=alphas_to_test, cv=kf, random_state=42))
lasso_pipe.fit(X_train, y_train)
lasso_preds = lasso_pipe.predict(X_test)
best_lasso_alpha = lasso_pipe.named_steps['lassocv'].alpha_

print("=== TASK 7: Ridge & Lasso Results ===")
print(f"Ridge | Best Alpha: {best_ridge_alpha} | Test MAE: {mean_absolute_error(y_test, ridge_preds):.3f} | Test RMSE: {np.sqrt(mean_squared_error(y_test, ridge_preds)):.3f} | Test R²: {r2_score(y_test, ridge_preds):.3f}")
print(f"Lasso | Best Alpha: {best_lasso_alpha} | Test MAE: {mean_absolute_error(y_test, lasso_preds):.3f} | Test RMSE: {np.sqrt(mean_squared_error(y_test, lasso_preds)):.3f} | Test R²: {r2_score(y_test, lasso_preds):.3f}")
```

```python
# Compare scaled coefficients across Linear, Ridge, and Lasso (Question 13)
scaled_lin = make_pipeline(StandardScaler(), LinearRegression()).fit(X_train, y_train)

coef_comparison = pd.DataFrame({
    "Feature": X.columns,
    "Linear (Scaled)": scaled_lin.named_steps['linearregression'].coef_.round(3),
    "Ridge": ridge_pipe.named_steps['ridgecv'].coef_.round(3),
    "Lasso": lasso_pipe.named_steps['lassocv'].coef_.round(3)
})
print("=== COEFFICIENT COMPARISON (SCALED) ===")
print(coef_comparison.to_string(index=False))
```

```python
print(" TASK 9: FINAL MODEL RECOMMENDATION ")
print("we have to choose Linear Regression (Degree 1)")
print("because of Lowest MAE (2.591), lowest RMSE (3.219), and highest R2 (0.906).")
print("also as additional Test on newer tickets and check consistency of complexity_score.\n")

print(" ANALYSIS QUESTIONS (10-17) ")
print("10. Mean Baseline: Sets a minimum benchmark (MAE = 9.114, R2 = -0.021) that real models must beat.")
print("11. MAE vs RMSE: Linear (Degree 1) won both (MAE = 2.591, RMSE = 3.219). RMSE is higher because it squares errors, punishing large outliers.")
print("12. Overfitting: Polynomial Degrees 2 and 3 made test RMSE worse (3.553 and 3.625 vs 3.219), proving they overfit the small training set.")
print("13. Regularization: Ridge and Lasso shrink feature weights toward zero to prevent overfitting while keeping test performance steady.")
print("14. Why Scale: Ridge and Lasso use one shared alpha penalty, so features must be scaled to the same range first to be penalized fairly.")
print("15. Data Leakage: lines_changed was removed from X because we only know how many lines changed AFTER a ticket is fixed.")
print("16. Split Strategy: Multi-year data needs a time-based split; repeated teams/customers need a grouped split (GroupKFold) so habits don't leak.")
print("17. Deployment Tests: Check errors across subgroups (reopened vs non-reopened), test on future tickets, and test sensitivity to complexity_score changes.")
```
